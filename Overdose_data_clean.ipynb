```python
import pandas as pd
import glob

# Where raw txt files were stored
RAW_PATH = "/Users/rosellazheng/Desktop/US_VitalStatistics/"
pattern = RAW_PATH + "Underlying Cause of Death, 20*.txt"

files = glob.glob(pattern)
print("Found files:", files)  # All are in there

# Combine all years
df = pd.concat((pd.read_csv(f, sep="\t") for f in files), ignore_index=True)

# Quick clean of column names
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

# 3. Save into project repo so everyone else can load this CSV
OUTPUT = "/Users/rosellazheng/Desktop/idsproject/vital_stats_2003_2015.csv"
df.to_csv(OUTPUT, index=False)

print("Saved clean CSV to:", OUTPUT)
```

```python
df_clean = df[df["notes"].isna()]  # this is droping all the weird notes
```

```python
print("Total rows after concat:", len(df_clean))  # 57241
```

```python
df_clean = df_clean.drop(columns=["notes", "year_code"])
```

```python
# here we want to split the county column and make a new one just for state
df_clean[["county", "state"]] = df_clean["county"].str.split(", ", expand=True)
```

```python
df_clean  # here we see its in the format we want
```

```python
print("Total rows after concat:", len(df_clean))  # 57241
```

```python
# extract states only used in research comparison
states_to_keepFL = ["FL", "GA", "SC", "NC"]
states_to_keepWA = ["WA", "OR", "MT", "CO"]
states_to_keep = states_to_keepFL + states_to_keepWA  # merge the lists
df_clean = df_clean[df_clean["state"].isin(states_to_keep)]
```

```python
# check for catorgories and make final dataset
df_clean["drug/alcohol_induced_cause"].unique()
```

```python
# categories to drop
drop_categories = [
    "All other alcohol-induced causes",
    "All other non-drug and non-alcohol causes",
    "Alcohol poisonings (overdose) (X45, X65, Y15)",
]

# keep everything EXCEPT the above
df_overdose = df_clean[~df_clean["drug/alcohol_induced_cause"].isin(drop_categories)]
```

```python
df_overdose
```

```python
print("Total rows after concat:", len(df_overdose))  # 2379
```

```python
df_overdose["deaths"] = pd.to_numeric(df_overdose["deaths"], errors="coerce").astype(
    "Int64"
)
df_overdose["county"] = df_overdose["county"].astype("string")
df_overdose["county_code"] = df_overdose["county_code"].astype("Int64")
df_overdose["drug/alcohol_induced_cause"] = df_overdose[
    "drug/alcohol_induced_cause"
].astype("string")
df_overdose["drug/alcohol_induced_cause_code"] = df_overdose[
    "drug/alcohol_induced_cause_code"
].astype("string")
df_overdose["year"] = df_overdose["year"].astype(int)
df_overdose["county_code"] = df_overdose["county_code"].astype(int)
df_overdose["state"] = df_overdose["state"].astype("string")
```
